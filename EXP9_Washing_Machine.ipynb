```python
import numpy as np
import skfuzzy as fuzzy
from skfuzzy import control as ctrl
```

```python
# 1. Define Input and Output Variables

# Inputs: Dirt and Grease (0 to 100%)
# Output: Wash Time (0 to 60 minutes)

dirt = ctrl.Antecedent(
    np.arange(0, 101, 1),
    'dirt'
)

grease = ctrl.Antecedent(
    np.arange(0, 101, 1),
    'grease'
)

wash_time = ctrl.Consequent(
    np.arange(0, 61, 1),
    'wash_time'
)
```

```python
# 2. Define Membership Functions for Dirt

dirt['small'] = fuzzy.trimf(
    dirt.universe,
    [0, 0, 50]
)

dirt['medium'] = fuzzy.trimf(
    dirt.universe,
    [0, 50, 100]
)

dirt['large'] = fuzzy.trimf(
    dirt.universe,
    [50, 100, 100]
)
```

```python
# Membership Functions for Grease

grease['small'] = fuzzy.trimf(
    grease.universe,
    [0, 0, 50]
)

grease['medium'] = fuzzy.trimf(
    grease.universe,
    [0, 50, 100]
)

grease['large'] = fuzzy.trimf(
    grease.universe,
    [50, 100, 100]
)
```

```python
# Membership Functions for Wash Time

wash_time['very_short'] = fuzzy.trimf(
    wash_time.universe,
    [0, 0, 15]
)

wash_time['short'] = fuzzy.trimf(
    wash_time.universe,
    [0, 15, 30]
)

wash_time['medium'] = fuzzy.trimf(
    wash_time.universe,
    [15, 30, 45]
)

wash_time['long'] = fuzzy.trimf(
    wash_time.universe,
    [30, 45, 60]
)

wash_time['very_long'] = fuzzy.trimf(
    wash_time.universe,
    [45, 60, 60]
)
```

```python
# 3. Define the Rule Base (9 Rules)

rule1 = ctrl.Rule(
    dirt['small'] & grease['small'],
    wash_time['very_short']
)

rule2 = ctrl.Rule(
    dirt['small'] & grease['medium'],
    wash_time['short']
)

rule3 = ctrl.Rule(
    dirt['small'] & grease['large'],
    wash_time['medium']
)

rule4 = ctrl.Rule(
    dirt['medium'] & grease['small'],
    wash_time['short']
)

rule5 = ctrl.Rule(
    dirt['medium'] & grease['medium'],
    wash_time['medium']
)

rule6 = ctrl.Rule(
    dirt['medium'] & grease['large'],
    wash_time['long']
)

rule7 = ctrl.Rule(
    dirt['large'] & grease['small'],
    wash_time['medium']
)

rule8 = ctrl.Rule(
    dirt['large'] & grease['medium'],
    wash_time['long']
)

rule9 = ctrl.Rule(
    dirt['large'] & grease['large'],
    wash_time['very_long']
)
```

```python
# 4. Control System Setup

wash_ctrl = ctrl.ControlSystem([
    rule1,
    rule2,
    rule3,
    rule4,
    rule5,
    rule6,
    rule7,
    rule8,
    rule9
])

washing_machine = ctrl.ControlSystemSimulation(
    wash_ctrl
)
```

```python
# 5. Test Case 1

washing_machine.input['dirt'] = 20
washing_machine.input['grease'] = 10

washing_machine.compute()

print(
    f"Inputs: Dirt=20%, Grease=10% -> "
    f"Calculated Wash Time: "
    f"{washing_machine.output['wash_time']:.2f} minutes"
)
```

```python
# Test Case 2

washing_machine.input['dirt'] = 80
washing_machine.input['grease'] = 85

washing_machine.compute()

print(
    f"Inputs: Dirt=80%, Grease=85% -> "
    f"Calculated Wash Time: "
    f"{washing_machine.output['wash_time']:.2f} minutes"
)
```

```python
# Additional Test Cases

test_cases = [
    (10, 10),
    (30, 40),
    (50, 50),
    (70, 60),
    (90, 90)
]

for dirt_value, grease_value in test_cases:

    washing_machine.input['dirt'] = dirt_value
    washing_machine.input['grease'] = grease_value

    washing_machine.compute()

    print(
        f"Dirt={dirt_value}% | "
        f"Grease={grease_value}% | "
        f"Wash Time={washing_machine.output['wash_time']:.2f} minutes"
    )
```

```python
import numpy as np
import skfuzzy as fuzzy
from skfuzzy import control as ctrl
```

```python
# 1. Define Input and Output Variables

# Inputs: Temperature and Humidity (0 to 100)
# Output: Fan Speed (0 to 100%)

temperature = ctrl.Antecedent(
    np.arange(0, 101, 1),
    'temperature'
)

humidity = ctrl.Antecedent(
    np.arange(0, 101, 1),
    'humidity'
)

fan_speed = ctrl.Consequent(
    np.arange(0, 101, 1),
    'fan_speed'
)
```

```python
# 2. Define Membership Functions for Temperature

temperature['low'] = fuzzy.trimf(
    temperature.universe,
    [0, 0, 50]
)

temperature['medium'] = fuzzy.trimf(
    temperature.universe,
    [0, 50, 100]
)

temperature['high'] = fuzzy.trimf(
    temperature.universe,
    [50, 100, 100]
)
```

```python
# Membership Functions for Humidity

humidity['low'] = fuzzy.trimf(
    humidity.universe,
    [0, 0, 50]
)

humidity['medium'] = fuzzy.trimf(
    humidity.universe,
    [0, 50, 100]
)

humidity['high'] = fuzzy.trimf(
    humidity.universe,
    [50, 100, 100]
)
```

```python
# Membership Functions for Fan Speed

fan_speed['low'] = fuzzy.trimf(
    fan_speed.universe,
    [0, 0, 50]
)

fan_speed['medium'] = fuzzy.trimf(
    fan_speed.universe,
    [0, 50, 100]
)

fan_speed['high'] = fuzzy.trimf(
    fan_speed.universe,
    [50, 100, 100]
)
```

```python
# 3. Define the Rule Base (9 Rules)

rule1 = ctrl.Rule(
    temperature['low'] & humidity['low'],
    fan_speed['low']
)

rule2 = ctrl.Rule(
    temperature['low'] & humidity['medium'],
    fan_speed['low']
)

rule3 = ctrl.Rule(
    temperature['low'] & humidity['high'],
    fan_speed['medium']
)

rule4 = ctrl.Rule(
    temperature['medium'] & humidity['low'],
    fan_speed['medium']
)

rule5 = ctrl.Rule(
    temperature['medium'] & humidity['medium'],
    fan_speed['medium']
)

rule6 = ctrl.Rule(
    temperature['medium'] & humidity['high'],
    fan_speed['high']
)

rule7 = ctrl.Rule(
    temperature['high'] & humidity['low'],
    fan_speed['medium']
)

rule8 = ctrl.Rule(
    temperature['high'] & humidity['medium'],
    fan_speed['high']
)

rule9 = ctrl.Rule(
    temperature['high'] & humidity['high'],
    fan_speed['high']
)
```

```python
# 4. Control System Setup and Defuzzification

fan_ctrl = ctrl.ControlSystem([
    rule1,
    rule2,
    rule3,
    rule4,
    rule5,
    rule6,
    rule7,
    rule8,
    rule9
])

fan_system = ctrl.ControlSystemSimulation(
    fan_ctrl
)
```

```python
# 5. Test Case 1

fan_system.input['temperature'] = 20
fan_system.input['humidity'] = 30

fan_system.compute()

print(
    f"Inputs: Temperature=20°C, Humidity=30% -> "
    f"Calculated Fan Speed: "
    f"{fan_system.output['fan_speed']:.2f}%"
)
```

```python
# Test Case 2

fan_system.input['temperature'] = 80
fan_system.input['humidity'] = 85

fan_system.compute()

print(
    f"Inputs: Temperature=80°C, Humidity=85% -> "
    f"Calculated Fan Speed: "
    f"{fan_system.output['fan_speed']:.2f}%"
)
```

```python
# Additional Test Cases

test_cases = [
    (10, 20),
    (30, 40),
    (50, 50),
    (70, 60),
    (90, 90)
]

for temp, hum in test_cases:

    fan_system.input['temperature'] = temp
    fan_system.input['humidity'] = hum

    fan_system.compute()

    print(
        f"Temperature={temp}°C | "
        f"Humidity={hum}% | "
        f"Fan Speed={fan_system.output['fan_speed']:.2f}%"
    )
```
